```python
import numpy as np

sensor_data = np.array([
    [12, 15, 14, 10, 13],
    [20, 22, 19, 18, 21],
    [30, 28, 27, 29, 31],
    [25, 24, 26, 23, 22]
])
Average = np.mean(sensor_data, axis=1)
print("Average per row:\n", Average )

Maximum = np.max(sensor_data, axis=0)
print("Maximun per column:\n", Maximum)

Normalized = sensor_data / Maximum
print("Normalized Values\n", Normalized)

for i in range(len(Average)):
  if Average[i] > 20:
    print("Machines", i+1)

Reshaped = sensor_data.reshape(2,10)
print("Reshaped Matrix:\n", Reshaped)
```

```python
import numpy as np

I = np.random.randint(1, 50, size=(3,3))
H = np.random.randint(1, 50, size=(3,3))
print("Matrix A:\n", I)
print("\nMatrix B:\n" , H)
M = I @ H
print("\nMultiplication of Matrices:\n" , M)
E = I * H
print("\nElement-wise Multiplication:\n", E)
greater_than_100 = M[M > 100]
print("\nElements greater than 100:\n", greater_than_100)
I[I < 10] = 0
H[H < 10] = 0
print("\nMatrix A after replacing <10 with 0:\n", I)
print("\nMatrix B after replacing <10 with 0:\n", H)
stacked = np.vstack((I, H))
print("\nVertically Stacked Matrix:\n", stacked)
```
