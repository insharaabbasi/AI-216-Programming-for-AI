```python
import pandas as pd

sales_data = {
    "OrderID": [101, 102, 103, 104, 105, 106],
    "Customer": ["Ali", "Sara", "Ahmed", "Ali", "Zain", "Sara"],
    "City": ["Karachi", "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Category": ["Electronics", "Clothing", "Electronics", "Groceries", "Clothing", "Electronics"],
    "Amount": [12000, 3500, None, 2400, 5200, 15000]
}

df = pd.DataFrame(sales_data)

df["Amount"] = df["Amount"].fillna(df["Amount"].median())
df["Amount"] = df["Amount"].astype(int)

df["HighValue"] = df["Amount"] > 10000
print("Data with HighValue column:")
print(df)

karachi_orders = df[df["City"] == "Karachi"]
print("\nOrders from Karachi:")
print(karachi_orders)

electronics_high = df[(df["Category"] == "Electronics") & (df["Amount"] > 10000)]
print("\nElectronics orders with Amount > 10000:")
print(electronics_high)

sorted_data = df.sort_values(by="Amount", ascending=False)
print("\nData sorted by Amount (descending):")
print(sorted_data)

top_customers = df.groupby("Customer")["Amount"].sum().sort_values(ascending=False).head(2)
print("\nTop 2 customers by total spending:")
print(top_customers)
```

```python
import pandas as pd

sales_data = {
    "OrderID": [101, 102, 103, 104, 105, 106],
    "Customer": ["Ali", "Sara", "Ahmed", "Ali", "Zain", "Sara"],
    "City": ["Karachi", "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Category": ["Electronics", "Clothing", "Electronics", "Groceries", "Clothing", "Electronics"],
    "Amount": [12000, 3500, None, 2400, 5200, 15000]
}

df = pd.DataFrame(sales_data)

df["Amount"] = df["Amount"].fillna(df["Amount"].median())
df["Amount"] = df["Amount"].astype(int)

total_revenue_city = df.groupby("City")["Amount"].sum()
print("Total revenue per city:")
print(total_revenue_city)

avg_category = df.groupby("Category")["Amount"].mean()
print("\nAverage order amount per category:")
print(avg_category)

unique_customers = df.groupby("City")["Customer"].nunique()
print("\nUnique customers per city:")
print(unique_customers)

pivot = pd.pivot_table(df, values="Amount", index="City", columns="Category", aggfunc="sum")
print("\nPivot Table (City vs Category):")
print(pivot)

revenue_df = df.groupby("City")["Amount"].sum().reset_index()
print("\nGrouped result with reset index:")
print(revenue_df)
```

```python
import pandas as pd

sales_data = {
    "OrderID": [101, 102, 103, 104, 105, 106],
    "Customer": ["Ali", "Sara", "Ahmed", "Ali", "Zain", "Sara"],
    "City": ["Karachi", "Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "Category": ["Electronics", "Clothing", "Electronics", "Groceries", "Clothing", "Electronics"],
    "Amount": [12000, 3500, None, 2400, 5200, 15000]
}

df = pd.DataFrame(sales_data)

# Fill missing Amount and convert to int
df["Amount"] = df["Amount"].fillna(df["Amount"].median())
df["Amount"] = df["Amount"].astype(int)

# Add new column TaxedAmount
df["TaxedAmount"] = df["Amount"] * 1.15

# Standardize City names to lowercase
df["City"] = df["City"].str.lower()

# Remove duplicate customers keeping highest purchase
df = df.sort_values("Amount", ascending=False).drop_duplicates("Customer", keep="first")

# Create summary dictionary
summary = {
    "total_orders": len(df),
    "total_revenue": df["Amount"].sum(),
    "average_order": df["Amount"].mean(),
    "top_city": df.groupby("City")["Amount"].sum().idxmax()
}

print("Processed DataFrame:")
print(df)

print("\nSummary Statistics:")
print(summary)

# Export final DataFrame to CSV
df.to_csv("final_sales_report.csv", index=False)
```
