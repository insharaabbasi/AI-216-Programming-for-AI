```python
sales_data = {
    "electronics": [
        ("Laptop", 120000, 3),
        ("Mouse", 1500, 10)
    ],
    "clothing": [
        ("Shirt", 2500, 5),
        ("Shoes", 5000, 2)
    ]
}

revenue_summary = {}
unique_products = set()

for category, products in sales_data.items():
    total_revenue = 0
    for name, price, quantity in products:
        total_revenue += price * quantity
        unique_products.add(name)
    revenue_summary[category] = total_revenue

for category, revenue in revenue_summary.items():
    print(f"Category: {category}")
    print(f"Total Revenue: {revenue}")
    print()

print("Unique Products Sold:")
for product in unique_products:
    print(product)
```

```python
retail_data = {
    "snacks": [
        ("cupcake", 50, 20),
        ("choclate", 100, 10),
        ("cocomo", 20, 30),
        ("lays", 100, 5),
        ("juice", 70, 7)
    ],
    "stationery": [
        ("notebook", 500, 5),
        ("highlighter", 150, 10),
        ("stickers", 100, 15),
        ("stickynotes", 200, 8)
    ],
    "books": [
        ("The Art of being Alone", 500, 8),
        ("Namal", 800, 5),
        ("You can be the happiest women in the world", 600, 3),
        ("The physocology of money", 700, 7)
    ]
}
def add_product(inventory, category, product_name, price, quantity):
    if category not in inventory:
        inventory[category] = []

    existing_names = {p[0] for p in inventory[category]}
    if product_name not in existing_names:
        inventory[category].append((product_name, price, quantity))
    else:
        print(f"Product '{product_name}' already exists in category '{category}'.")

def update_quantity(inventory, category, product_name, added_quantity):
    if category in inventory:
        new_list = []
        for p in inventory[category]:
            if p[0] == product_name:

                new_quantity = p[2] + added_quantity
                new_list.append((p[0], p[1], new_quantity))
            else:
                new_list.append(p)
        inventory[category] = new_list

def remove_sold_out(inventory):
    for category in inventory:
        inventory[category] = [p for p in inventory[category] if p[2] > 0]

def get_unique_products(inventory):
    unique_set = set()
    for products in inventory.values():
        for p in products:
            unique_set.add(p[0])
    return unique_set

def generate_report(inventory):
    report = {}
    for category, products in inventory.items():
        total_value = sum(p[1] * p[2] for p in products)
        report[category] = total_value
    return report

add_product(retail_data, "snacks", "cookies", 60, 12)
add_product(retail_data, "electronics", "headphones", 1500, 5)


update_quantity(retail_data, "snacks", "cupcake", 5)
update_quantity(retail_data, "books", "Namal", -5)

remove_sold_out(retail_data)

unique_products = get_unique_products(retail_data)

report = generate_report(retail_data)

print("FINAL INVENTORY")
for category, products in retail_data.items():
    print(category, ":", products)

print("UNIQUE PRODUCTS")
print(unique_products)

print("INVENTORY VALUE REPORT")
print(report)

```
