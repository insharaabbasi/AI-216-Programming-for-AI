```python
from sklearn.datasets import load_wine
import pandas as pd

wine = load_wine()

df = pd.DataFrame(wine.data, columns=wine.feature_names)

df['target'] = wine.target

print(df.head())

print("Shape:", df.shape)

print(df['target'].value_counts())
```

```python
X = df.drop('target', axis=1)
y = df['target']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,   # you can also try 0.25 later
    random_state=42
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(max_iter=5000)
lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)

knn_pred = knn_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

lr_acc = accuracy_score(y_test, lr_pred)
knn_acc = accuracy_score(y_test, knn_pred)

print("Logistic Regression Accuracy:", lr_acc)
print("KNN Accuracy:", knn_acc)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, lr_pred)
print("Confusion Matrix (Logistic Regression):\n", cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, lr_pred))
```

```python
X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X, y,
    test_size=0.5,
    random_state=42
)

model2 = LogisticRegression(max_iter=5000)
model2.fit(X_train2, y_train2)

pred2 = model2.predict(X_test2)

print("Accuracy (50% test):", accuracy_score(y_test2, pred2))
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

def ml_pipeline(model, name):
    model.fit(X_train, y_train)
    pred = model.predict(X_test)

    print("\nModel:", name)
    print("Accuracy:", accuracy_score(y_test, pred))
    print("Confusion Matrix:\n", confusion_matrix(y_test, pred))


ml_pipeline(LogisticRegression(max_iter=5000), "Logistic Regression")
ml_pipeline(KNeighborsClassifier(n_neighbors=5), "KNN")
```
