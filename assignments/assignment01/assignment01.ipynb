```python
attendence = [72, 85, 90, 60, 40, 78, 88]
eligible_count = 0
noteligible_count = 0

for a in attendence:
  if a > 75:
        eligible_count += 1
  else:
        noteligible_count +=1

print(f"Eligible Students:", eligible_count)
print(f"Not Eligible Students:" , noteligible_count)
```

```python
CGPA = [3.6, 3.9, 3.6, 3.76, 3.49, 3.45, 3.5]
Income = [40000, 80000, 70000, 90000, 60000, 30000, 780000]
eligible = 0

for i in range (len(CGPA)):
  if CGPA[i] >= 3.5 and Income[i] <= 80000:
    eligible = eligible + 1

print("Eligible for Scholarship:", eligible )
```

```python
readings = [20, 40, 60, 80, 10, 30, 50, 100, 90, 70]
alerts = 0

for r in readings:
  if r < 20 or r > 80:
    alerts = alerts + 1
print("Total Alerts are: ", alerts)

average = (alerts/len(readings)) * 100

print("The average of alerts is:", average)

```
