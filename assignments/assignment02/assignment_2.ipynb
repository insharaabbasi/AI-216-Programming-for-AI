```python
students_data = [
    {"id": "S01", "name": "Ali", "scores": (78, 85, 90)},
    {"id": "S02", "name": "Sara", "scores": (88, 92, 79)},
    {"id": "S03", "name": "Ahmed", "scores": (45, 60, 55)},
    {"id": "S04", "name": "Zain", "scores": (95, 91, 89)}
]

def calculate_average(scores):
    return sum(scores) / len(scores)

def determine_grade(average):
    if average >= 85:
        return "A"
    elif 70 <= average <= 84:
        return "B"
    elif 50 <= average <= 69:
        return "C"
    else:
        return "F"

def top_student(students):
    best_student = None
    highest_average = 0
    for student in students:
        if student["average"] > highest_average:
            best_student = student
            highest_average = student["average"]
    return best_student

def passed_students(students):
    passed = []
    for student in students:
        if all(score >= 50 for score in student["scores"]):
            passed.append(student)
    return passed

for student in students_data:
    student["average"] = calculate_average(student["scores"])
    student["grade"] = determine_grade(student["average"])

print("Student Performance:")
for student in students_data:
    print(student["name"], "- Average:", round(student["average"], 2),
          "Grade:", student["grade"])

print("Top Student:", top_student(students_data)["name"])
print("Passed Students:", [student["name"] for student in passed_students(students_data)])

class StudentAnalytics:
    def __init__(self, students_data):
        self.students_data = students_data

    def compute_results(self):
        for student in self.students_data:
            avg = sum(student["scores"]) / len(student["scores"])
            student["average"] = avg
            if avg >= 85:
                student["grade"] = "A"
            elif avg >= 70:
                student["grade"] = "B"
            elif avg >= 50:
                student["grade"] = "C"
            else:
                student["grade"] = "Fail"

    def get_top_student(self):
        top = self.students_data[0]
        for student in self.students_data:
            if student["average"] > top["average"]:
                top = student
        return top

    def get_class_average(self):
        total = sum(student["average"] for student in self.students_data)
        return total / len(self.students_data)

    def get_unique_grades(self):
        grades = set(student["grade"] for student in self.students_data)
        return grades

    def generate_report(self):
        distribution = {}
        for student in self.students_data:
            grade = student["grade"]
            distribution[grade] = distribution.get(grade, 0) + 1
        return {
            "class_average": self.get_class_average(),
            "top_student": self.get_top_student()["name"],
            "grade_distribution": distribution
        }

    def improving_students(self):
        improved = []
        for student in self.students_data:
            s = student["scores"]
            if s[0] < s[1] and s[1] < s[2]:
                improved.append(student["name"])
        return improved

    def to_table(self):
        table = {"ids": [], "names": [], "averages": [], "grades": []}
        for student in self.students_data:
            table["ids"].append(student["id"])
            table["names"].append(student["name"])
            table["averages"].append(student["average"])
            table["grades"].append(student["grade"])
        return table

analytics = StudentAnalytics(students_data)
analytics.compute_results()

print("\nReport:", analytics.generate_report())
print("Improving Students:", analytics.improving_students())
print("Tabular Format:", analytics.to_table())
```

```python
courses_data = {
    "Python": {
        "instructor": "Dr. A",
        "students": {"S01", "S02", "S03"}
    },
    "Machine Learning": {
        "instructor": "Dr. B",
        "students": {"S02", "S04"}
    },
    "Data Science": {
        "instructor": "Dr. C",
        "students": {"S01", "S04"}
    }
}

def students_in_multiple_courses(courses):
    student_count = {}
    for course in courses.values():
        for student in course["students"]:
            student_count[student] = student_count.get(student, 0) + 1
    multiple_courses = [s for s, count in student_count.items() if count > 1]
    return multiple_courses

def courses_with_many_students(courses):
    result = []
    for course_name, info in courses.items():
        if len(info["students"]) > 2:
            result.append(course_name)
    return result

def student_course_mapping(courses):
    mapping = {}
    for course in courses.values():
        for student in course["students"]:
            mapping[student] = mapping.get(student, 0) + 1
    return mapping

def all_unique_students(courses):
    students_set = set()
    for course in courses.values():
        students_set.update(course["students"])
    return students_set

print("Students in more than one course:", students_in_multiple_courses(courses_data))
print("Courses with more than 2 students:", courses_with_many_students(courses_data))
print("Student → Number of Courses Mapping:", student_course_mapping(courses_data))
print("All Unique Students:", all_unique_students(courses_data))
```
