```python
# SMART PLANTING ASSISTANT
import pandas as pd

data = {
    "plant": [
        "mint","tomato","coriander","aloe vera","spinach",
        "chili","rose","sunflower","neem tree","money plant",
        "jasmine","mango tree","lavender","basil","marigold",
        "tulip","strawberry","cactus","wheat grass","hibiscus"
    ],
    "season": [
        "spring","summer","winter","all","winter",
        "summer","spring","summer","all","all",
        "spring","summer","spring","spring","autumn",
        "winter","winter","summer","all","summer"
    ],
    "region": [
        "all","punjab","all","all","punjab",
        "punjab","punjab","punjab","all","all",
        "sindh","sindh","balochistan","kpk","punjab",
        "kashmir","kpk","balochistan","all","punjab"
    ],
    "sunlight": [
        "partial","full","partial","full","full",
        "full","full","full","full","partial",
        "full","full","full","partial","full",
        "partial","full","full","partial","full"
    ],
    "water": [
        "medium","high","medium","low","medium",
        "medium","medium","low","low","medium",
        "medium","high","low","medium","low",
        "medium","medium","low","high","medium"
    ],
    "growth_days": [
        35,75,28,90,45,
        80,120,85,365,50,
        60,300,80,40,60,
        130,90,250,10,75
    ],
    "soil": [
        "moist","loamy","dry","sandy","loamy",
        "loamy","loamy","dry","any","moist",
        "loamy","loamy","sandy","moist","dry",
        "well-drained","loamy","sandy","loamy","loamy"
    ],
    "space": [
        "indoor","outdoor","indoor","indoor","outdoor",
        "outdoor","garden","garden","garden","indoor",
        "balcony","garden","balcony","indoor","balcony",
        "outdoor","outdoor","outdoor","indoor","garden"
    ],
    "type": [
        "herb","vegetable","herb","succulent","vegetable",
        "vegetable","flower","flower","tree","indoor plant",
        "flower","tree","flower","herb","flower",
        "flower","fruit","succulent","grass","flower"
    ]
}

df = pd.DataFrame(data)

print("WELCOME TO SMART PLANTING ASSISTANT")
print("YOUR PLANT BUDDY IS HERE!let’s find the perfect match for your space")

seasons = ["spring","summer","autumn","winter","all"]
regions = ["punjab","sindh","kpk","balochistan","kashmir","all"]
spaces = ["indoor","outdoor","balcony","garden","terrace"]
types = ["herb","vegetable","flower","tree","fruit","succulent","indoor plant"]

def get_input(prompt, options):
    while True:
        print(f"\n {prompt}")
        print("Options:", ",".join(options))
        user = input("Enter Here:").strip().lower()
        if user in options:
            return user
        print("Invalid input, try again.")

user_season = get_input("Select Season", seasons)
user_region = get_input("Select Region", regions)
user_space = get_input("Select Space", spaces)
user_type = get_input("Select Plant Type", types)

print("\n Weather (hot / cold / humid / dry / mild)")
user_weather = input("Enter Here:").strip().lower()

filtered = df[
    (df["season"].isin([user_season, "all"])) &
    (df["region"].isin([user_region, "all"])) &
    (df["space"] == user_space) &
    (df["type"] == user_type)
]

top = filtered.head(5)

print("\n TOP PLANT RECOMMENDATIONS FOR YOU:\n")

if top.empty:
    print(" No matches found. Try adjusting your choices your perfect plant is waiting somewhere")
else:
    for i, r in enumerate(top.itertuples(), 1):
        print(f"{i}. {r.plant.title()} ({r.type})")
        print(f"Season: {r.season}")
        print(f"Region: {r.region}")
        print(f"Sunlight: {r.sunlight}")
        print(f"Water: {r.water}")
        print(f"Soil: {r.soil}")
        print(f"Space: {r.space}")
        print(f"Growth Time: {r.growth_days} days")
        print("-" * 45)

print("\n IMPORTANT INFORMATION FOR YOU")

if not top.empty:
    print("Average growth time:", top["growth_days"].mean())
    print("Fastest plant:", top.loc[top["growth_days"].idxmin(), "plant"])

print("\n ESSENTIAL GARDENING TOOLS:\n")

tools = [
    "Hand Trowel","Spade","Shovel","Garden Fork",
    "Hoe","Loppers","Soil Scoop","Dibber",
    "Shears","Weeder","Sickle","Watering Can"
]

for t in tools:
    print("-", t)

print("\n SMART TIPS:")

if user_space in ["indoor","balcony"]:
    print("- Use drainage pots")
if user_weather == "hot":
    print("- Water early morning")
elif user_weather == "cold":
    print("- Protect from frost")
elif user_weather == "humid":
    print("- Ensure airflow")
elif user_weather == "dry":
    print("- Maintain soil moisture")

print("\n Just a heads-up: I'm still learning double check anything you're unsure about")
print("\n Happy Gardening!")
```
